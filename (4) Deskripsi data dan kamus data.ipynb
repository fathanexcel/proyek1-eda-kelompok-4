# Deskripsi Data dan Kamus Data

* **Sumber data beserta link dan lisensinya:** Dataset dari **https://www.kaggle.com/datasets/sehaj1104/indian-road-accident-dataset-20222025/data** dan lisensinya adalah **CC0: Public Domain**.
* **Siapa yang mengumpulkan data, kapan, dan bagaimana:**
  1. Pengumpul data: Mansehaj Preet
  2. Kapan: 2022–2025 (Update: Maret 2026)
  3. Bagaimana: Dataset ini dihasilkan menggunakan pendekatan simulasi berbasis aturan yang dikombinasikan dengan teknik pengambilan sampel acak. Fitur-fitur utama seperti kepadatan lalu lintas, kondisi cuaca, dan tingkat keparahan kecelakaan ditentukan menggunakan distribusi probabilistik untuk mencerminkan skenario yang realistis. Koordinat geospasial dihasilkan dalam batas-batas kota yang telah ditetapkan sebelumnya. Fitur temporal seperti jam sibuk dan akhir pekan diturunkan dari nilai tanggal dan waktu. Fitur kontekstual tambahan, seperti indikator festival, disertakan dengan menggunakan referensi kalender nyata. Skor risiko gabungan disusun berdasarkan faktor-faktor yang berkontribusi—seperti lalu lintas, jarak pandang, cuaca, dan waktu kejadian—guna meningkatkan kegunaan dataset bagi aplikasi pembelajaran mesin.
* **Jumlah baris dan kolom:**
  1. Total sebenarnya: **20.000** baris, dan **24** kolom
  2. Sampel: **2.000** baris teratas dan **24** kolom
