```python
import pandas as pd
df = pd.read_csv("dataset_bersih.csv")

kolom_kategorik = 'road_type'

frekuensi = df[kolom_kategorik].value_counts()
persentase = df[kolom_kategorik].value_counts(normalize=True) * 100

tabel = pd.DataFrame({
    'Frekuensi': frekuensi,
    'Persentase (%)': persentase.round(2)
})

modus = df[kolom_kategorik].mode()[0]

print(f"VARIABEL KATEGORIK")
display(tabel)
print(f"\nModus (Kategori terbanyak): {modus}")
```
