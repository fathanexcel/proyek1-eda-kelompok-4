```python
import pandas as pd
df = pd.read_csv("dataset_bersih.csv")
kp = ['vehicles_involved', 'casualties', 'risk_score']

hasil = []

for nk in kp:
    kolom = df[nk]
    
    hasil.append({
        'Variabel Numerik': nk,
        'Mean': kolom.mean(), #bisa ditambah round biar bisa berapa angka dibelakang koma
        'Median': kolom.median(),
        'Modus': kolom.mode()[0],
        'Nilai Min': kolom.min(),
        'Nilai Max': kolom.max(),
        'Range': kolom.max() - kolom.min(),
        'Varians': round(kolom.var(), 2),
        'Standar Deviasi': round(kolom.std(), 2),
        'Quantil (0,25)': kolom.quantile(0.25),
        'Quantil (0,75)': kolom.quantile(0.75),
        'IQR': kolom.quantile(0.75) - kolom.quantile(0.25)
    })

tabel = pd.DataFrame(hasil)
tabel
```
