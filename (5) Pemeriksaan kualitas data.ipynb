```python
print("""
TEMUAN:
1. Tipe data kolom 'date' dan 'time' masih berupa teks (str), seharusnya datetime.
2. Kolom 'festival' memiliki 1994 baris data yang kosong (kosong hampir 100%).
3. Kolom 'hour' dan 'time' isinya adalah sama, sehingga kolom 'hour' tidak berguna (tidak memberikan informasi tambahan apapun).
4. Tidak ada data yang duplikat (0).

KEPUTUSAN:
1. Ubah kolom 'date' dan 'time' menjadi tipe datetime agar waktu bisa dianalisis.
2. Buang (drop) seluruh kolom 'festival' karena mayoritas datanya kosong dan tidak berguna untuk analisis.
3. Buang (drop) seluruh kolom 'hour' karena datanya sama seperti kolom 'time'.
3. Sisa kolom lainnya dibiarkan karena sudah bersih.
""")
```

```python
import pandas as pd

df = pd.read_csv("indian_roads_dataset.csv", nrows = 2000)

df['date'] = pd.to_datetime(df['date'])
df['time'] = pd.to_datetime(df['time'], format='%H:%M').dt.time

df = df.drop(columns=['festival', 'hour'])
df
```

```python
df.to_csv("dataset_bersih.csv", index=False)
```
