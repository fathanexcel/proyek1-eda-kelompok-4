```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("dataset_bersih.csv")

kolom_list = ["vehicles_involved", "casualties"]

for kolom in kolom_list:
    plt.figure(figsize=(7, 4))
    plt.hist(df[kolom].dropna(), bins=5, edgecolor="black", color="steelblue")
    plt.title(f"Histogram {kolom}")
    plt.xlabel(kolom)
    plt.ylabel("Frekuensi")
    plt.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()

print("Secara logika semakin banyak kendaraan yang terlibat dalam kecelakaan maka semakin banyak pula korban kecelakaan namun berdasarkan tabel histogram vehicle_involved distribusinya hampir simetris sedangkan histogram casualities distribusinya miring kekanan hal ini dikarenakan banyaknya faktor yang mempengaruhi jumlah korban diluar dari kendaraan yang terlibat")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("dataset_bersih.csv")

plt.boxplot(df["risk_score"].dropna(), patch_artist=True, boxprops=dict(facecolor="steelblue"))
plt.title("Boxplot Risk Score")
plt.ylabel("Risk Score")
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()

print("Berdasarkan Boxplot, Risk Score menunjukkan sebaran data yang terkonsentrasi pada 0.25 - 0.6. Nilai median berada disekitar (0.25 - 0.6) . Data tidak memiliki outlier yang menunjukkan bahwa data memiliki batas nilai yang jelas dan distribusi yang merata. Maka berdasarkan sebaran data tersebut diperoleh bahwa sebagian besar insiden kecelakaan di jalan raya memiliki skor risiko yang berada di kisaran menengah kebawah yaitu 0.25 - 0.6")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("dataset_bersih.csv")

df_grouped = (df.groupby("road_type")["casualties"].size().reset_index().head(3))

x = df_grouped["road_type"]
y = df_grouped["casualties"]

plt.figure(figsize=(25, 9))
plt.bar(x, y, color="steelblue", edgecolor="black")
plt.ylim(600, 700)
plt.title("Road Type VS Casualties", fontsize=17)
plt.ylabel("Casualties", fontsize=17)
plt.yticks(fontsize=17)
plt.xlabel("Road Type", fontsize=17)
plt.xticks(fontsize=17)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

print("Berdasarkan data tersebut diperoleh bahwa road type urban memiliki korban terbanyak disusul dengan rural lalu highway")
```
