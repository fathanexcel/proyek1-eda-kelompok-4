## Latar belakang dan pertanyaan

Dataset *Indian Roads* ini menarik karena tidak sekadar menyalahkan kelalaian pengemudi saat terjadi kecelakaan, tetapi merekam kondisi di lapangan secara lebih luas, mulai dari cuaca saat kejadian, tingkat kemacetan, ada tidaknya lampu lalu lintas, hingga jenis jalan yang dilewati. Informasi ini mempermudah kita untuk membongkar akar penyebab kecelakaan yang mungkin terabaikan. Hasil temuan dari data ini relevan untuk mewujudkan konsep *Smart City*, di mana pemerintah bisa memperbaiki fasilitas dan tata letak jalan raya sedini mungkin agar warganya bisa berkendara dengan jauh lebih aman.

**Pertanyaan analisis:**
1. Berapa rata-rata, nilai tengah, dan modus dari variabel *vehicles_involved*, *casualties*, serta *risk_score* pada sampel data kecelakaan ini?
2. Seberapa besar tingkat persebaran data ketiga variabel tersebut berdasarkan nilai standar deviasi, range, dan IQR (Interquartile Range)?
3. Bagaimana perbandingan antara rata-rata (*mean*) dan nilai tengah (*median*) pada masing-masing variabel, serta apa dampaknya terhadap kemiringan sebaran data?
